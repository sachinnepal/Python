```python
import numpy as np
import pandas as pd
```

```python
#string

country=['nepal','india','pakistan']
pd.Series(country)
```

```python
#integer

numbers=[70,80,90,]
pd.Series(numbers)
```

```python
#custom index

numbers=[70,80,90,]
subjects=['maths','science','english']
pd.Series(numbers,index=subjects)
```

```python
#setting a name

marks=pd.Series(numbers,index=subjects,name='marks')
marks
```

```python
#series from dict
marks={
    'maths':67,
    'science':80
}
marks_series=pd.Series(marks,name='Marks')
marks_series
```

Series Attributes

```python
#size
marks_series.size
```
